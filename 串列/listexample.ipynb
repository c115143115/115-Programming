## 走訪串列的三種寫法

```python
print(scores)
for i in range(len(scores)):
    print(scores[i])
```

```python
# 對於 scores 中的每個元素,一次取一個,放在 e 中
for e in scores:
    print(e)
```

```python
for i, e in enumerate(scores):
    print(i,e)
```
